```python
# 1. GitHub 인증 토큰을 사용하여 원격 저장소 주소를 설정
from google.colab import userdata

GH_TOKEN = userdata.get("GH_TOKEN")

# 기존에 등록된 origin이 있다면 제거하고 새로 등록
!git remote remove origin 2>/dev/null
!git remote add origin https://younghong:{GH_TOKEN}@github.com/youn9hon9/PSPY-DataAnalysis.git
print("원격 저장소(origin) 등록 완료!")
```

```python
# 2. 원격 저장소의 최신 버전(origin/main)으로 로컬 파일들을 완벽하게 덮어쓰고 동기화
# (주의: 로컬에서 임의로 수정한 내역이 있다면 모두 사라지고 GitHub의 깔끔한 최신 상태가 됩니다)
!git fetch origin main
!git reset --hard origin/main
```

```python
# 3. 동기화된 최신 커밋 메시지만 확인
!git log -1 --date=format:'%Y-%m-%d %H:%M:%S' --pretty=format:"[%ad] %an - %s"
```
